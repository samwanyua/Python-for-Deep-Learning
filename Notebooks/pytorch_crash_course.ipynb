## Tensor Operations

```python
x = torch.ones(2,2)
y = torch.rand(2,2)

# addition
sum = x + y
print("sum",sum)
print(torch.add(x,y)) # a cleaner way

# subtraction
print("subtraction", torch.sub(x,y))

# multiplication
print("multiplication", torch.mul(x,y))

# division
print("division", torch.div(x,y))

# inplace addition
print("inplace addition",y.add_(x))

```

### AutoGrad
Autograd package provide automatic differentiation for operations on Tensors.
- torch.autograd is an engine for computing the vector-Jacobian product. It computes partial derivatives while applying the chain rule

```python
t = torch.rand(3, requires_grad=True)
u = t + 2
print(t)
print(u) # u was created as a result of an operation, it has grad_fn attribute
# grad_fn -> references a function that has created the tensor

z = u * u * 3
print("z: ",z)
z = z.mean()
print("mean of z: ", z.mean())
print(u.grad)
z.backward() # calculate the gradient with backpropagation
print(u.grad)
# print("gradient of x: ", t.grad)
```

 ## Stop a tensor from tracking history
- Ex. during the training loop when we want to update weights or after training during evaluation
- to prevent that we can use;
- x.requires_grad(False)
- x.detach()
- wrap in wtih torch.no_grad()

```python
a = torch.randn(3,3) # randn -> standard normal dist. mean = 0, standard deviation = 1, range -infinity to +infinity : value cluster around 0
# rand() => uniform distribution, range[0,1]
b = (a * a).sum()
print(a, a.dtype)
print(a.requires_grad)
print(b.grad_fn)

a.requires_grad_(True)
b = (a * a).sum()
print(a.requires_grad)
print(b.grad_fn)
```

```python
#.detach()
a = torch.randn(3,3, requires_grad=True)
b = a.detach()
print(a.requires_grad)
print(b.requires_grad)
```

```python
# wrap in with 'torch.no_grad()'
a = torch.randn(2,2, requires_grad=True)
print(a.requires_grad)
with torch.no_grad():
  b = a**2
  print(b.requires_grad)
```

## Gradient Descent Autograd
Linear regression: f(x) = w*x +b
- where w = weights, b = bias


```python
import torch
# f = w * x +b
x = torch.tensor([1,2,3,4,5,6,7,8], dtype=torch.float32)
y = torch.tensor([2,4,6,8,10,12,14,16], dtype=torch.float32)

w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

# model output
def forward(x):
  return w * x

# loss = MSE
def loss(y, y_pred):
  return ((y_pred - y)**2).mean()

x_test = 5.0

print(f'Prediction before training: f({x_test}) = {forward(x_test).item():.3f}')
```

```python
# Training
learning_rate = 0.01
n_epochs = 100

for epoch in range(n_epochs):
  # predict = forward pass
  y_pred = forward(x)
  # loss
  l = loss(y, y_pred)
  # calculate gradients = backward pass
  l.backward()

  # update weights
  # w.data = w.data - learning)rate * w.grad
  with torch.no_grad():
    w -= learning_rate * w.grad # gradient descent
  # zero the gradients after updating
  w.grad.zero_()

  if (epoch+1) % 10 == 0:
    print(f'epoch {epoch+1}: w = {w.item():.3f}, loss = {l.item():.3f}')

print(f'Prediction after training: f({x_test}) = {forward(x_test).item():.3f}')
```

### Model, Loss and Optimizer
A typical pytorch pipeline looks like:
* Design model: input, output, forward pass with different layers
* Construct loss optimizer
* Training loop: forward (compute predictin and loss), backward(compute gradients), update weights



```python
import torch
import torch.nn as nn

# Training data
x = torch.tensor([[1],[2],[3],[4],[5],[6],[7],[8]], dtype=torch.float32)
y = torch.tensor([[2],[4],[6],[8],[10],[12],[14],[16]], dtype=torch.float32)
print(x.shape)
n_samples, n_features = x.shape
print(f"samples: {n_samples}, features: {n_features}")

# Design the model
class LinearRegression(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(LinearRegression, self).__init__()
        self.lin = nn.Linear(input_dim, output_dim)

    def forward(self, x):
        return self.lin(x)

input_size, output_size = n_features, n_features
model = LinearRegression(input_size, output_size)

# Define a test input as a tensor
x_test = torch.tensor([[4.0]], dtype=torch.float32)

# Prediction before training
print(f'Prediction before training: f({x_test.item()}) = {model(x_test).item():.3f}')

# Define loss and optimizer
learning_rate = 0.01
n_epochs = 100

loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Training loop
for epoch in range(n_epochs):
    # Forward pass
    y_pred = model(x)
    # Loss
    l = loss_fn(y, y_pred)
    # Backpropagation
    l.backward()
    # Update weights
    optimizer.step()
    # Zero gradients
    optimizer.zero_grad()

    if (epoch+1) % 10 == 0:
        w, b = model.parameters()
        print(f'epoch {epoch+1} \n w = {w[0][0].item()} \n loss = {l.item()}')

# Prediction after training
print(f'Prediction after training: f({x_test.item()}) = {model(x_test).item():.3f}')
```
